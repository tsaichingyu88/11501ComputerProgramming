## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？Python 使用任意精度整數（Arbitrary Precision Integers）、整數以可變長度的陣列存儲，不受固定位元數限制、根據需要動態分配記憶體。
2. Python 的 `int` 是否具有固定的位元數限制？否。只要有足夠記憶體，可以表示任意大的整數。
3. 什麼是 Arbitrary Precision（任意精度）？能夠精確表示任意大小的整數，不因位元寬度限制而產生溢出、CPython 自動管理位元數，使用者無需干預。
4. CPython 使用什麼物件來實作 `int`？PyLongObject 結構體（定義於 Include/longobject.h）。
5. Python 整數大小真正的限制是什麼？可用的系統記憶體（RAM）、理論上沒有上限，實務上受限於電腦硬體。

---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？浮點數採用 IEEE 754 標準，使用二進制表示、0.1、0.2 等十進制小數無法精確轉換成二進制、計算過程中不斷累積舍入誤差。
2. 浮點數是否能精確表示所有小數？否。只有特定的二進制小數能被精確表示。
3. 什麼是浮點數的精度（precision）？指能夠精確表示的有效位數。
4. 什麼是浮點數的範圍（range）？指能表示的最小正數到最大正數。
5. 精度與範圍有何不同？兩者獨立但互相制約。

## 第八部分：綜合思考

### 思考題一

請回答：

1. Unlimited Precision 是什麼意思？無限制精度，整數能精確表示任意大的值，不受位元寬度限制。
2. 它是否代表可以儲存無限大的整數？理論上是（在記憶體允許的範圍內），實務上受限於可用 RAM。

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？採用可變長度陣列儲存、根據數值大小自動擴展、只要有記憶體就能支援更大的數。
2. 為什麼 `float` 卻存在範圍限制？IEEE 754 標準規定了固定的位元配置、指數位數固定（FP64 是 11 位）、無法超越這個設計。
3. 為什麼 `float` 可能產生誤差？二進制無法精確表示十進制小數、舍入誤差累積、不同運算順序導致結果不同。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。
Python：正確性優先，容許使用更多記憶體和時間。
AI：效率優先，容許接受小的精度損失。
2. 各自的優點與缺點。
Python int
優點： 無誤差，結果完全精確、無限精度，適合密碼學、金融。
缺點：記憶體開銷大、 運算速度慢。
FP16/BF16
優點：節省記憶體（50% 相比 FP32）、極快運算速度（硬體支援）。
缺點：精度低（3-4 位十進制）、可能數值不穩定。
3. 為什麼大型語言模型通常選擇較低精度的表示方式？模型龐大、計算密集、硬體最佳化、神經網路容錯性。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？記憶體爆炸、訓練時間加倍、GPU 記憶體不足、成本激增、不必要浪費。

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。

角度一：記憶體使用量
10**1000 的 Python int：
- 1000 位十進制數 ≈ 3322 位二進制
- 使用 30-bit digits：3322 / 30 ≈ 111 個 digits
- 每個 digit 8 bytes + 物件開銷 ≈ 1-2 KB

GPT-3 (1750 億參數):
- 若用 Python int：1.75×10¹¹ × 1-2 KB ≈ 175-350 TB ❌ 不可能
- 若用 FP32：1.75×10¹¹ × 4 bytes = 700 GB（需要 GPU）
- 若用 FP16：1.75×10¹¹ × 2 bytes = 350 GB（可行）

角度二：計算速度
操作 10**1000：
- Python int 加法：大數運算複雜度 O(n)，n 是位數

AI 模型單次前向傳播（GPT-3）：
- 1750 億 × 2000 token 位置 ≈ 3.5×10¹⁴ 次浮點運算
- FP32：在 NVIDIA H100（1.46 TFLOPS）上需要 ~240 秒
- Python int：會慢 1000+ 倍 ❌

硬體最佳化：
- GPU 每秒執行 10¹² 次 FP16 運算
- GPU 對整數運算無特殊優化

角度三：數值精度
精度實際需求：

Python int：
- 需要完全精確（金融、密碼學等）
- 一位錯誤導致完全失敗

神經網路：
- 3-4 位精度足夠（FP16 精度）
- 模型本身對權重變化有容錯性
- 訓練過程中有足夠的噪聲

實驗數據：
- 用 FP16 訓練的 GPT-2 vs FP32：
  準確率差異 < 0.5%（可接受）
- 如果用 Python int：速度慢 1000 倍（不可接受）

角度四：GPU 運算效率
硬體架構設計：

GPU 強項：
✓ FP32/FP16 有專用計算單元
✓ 支援張量核心（Tensor Cores）
✓ NVIDIA H100：每秒 1.46 TFLOPS （FP16）

GPU 弱項：
✗ 整數運算無特殊優化
✗ 任意精度運算無硬體支援
✗ Python 物件開銷太大

TPU (Google Tensor Processing Unit)：
✓ 對 BF16 特別優化
✓ 支援混合精度訓練
✗ 對 FP64 支援較差
